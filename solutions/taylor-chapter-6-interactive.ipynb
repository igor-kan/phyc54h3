## 6.1 
In spherical polar coordinates a sphere of radius $R$ is $S=\set{(r,\theta, \phi) \in \mathbb{R}^{3} | r = R, \theta \in [0,\pi], \phi \in [0, 2\pi)}$ -- this is probably not a proper notation. 

The sphere is paremetrized by colatitude(or meridional angle) $\theta$ and azimuth(or azimuthal) $\phi$
$\mathbf{r}(\theta, \phi) = \begin{bmatrix} R \sin\theta \cos\phi \\ R \sin\theta \sin\phi \\ R \cos\theta \end{bmatrix}$

$$d\mathbf{r} = \left(\frac{\partial \mathbf{r}}{\partial \theta}\right) d\theta + \left(\frac{\partial \mathbf{r}}{\partial \phi}\right) d\phi$$
$$ds = \Vert{}d\mathbf{r}\Vert{} = \sqrt{R^2 \, d\theta^2 + R^2 \sin^2\theta \, d\phi^2}$$

Simple intuitive derivation with the Pythagorean theorem with hypotenuse $ds$, and visual proof will probably suffice. 


We need to show that the length of the path between any two points $(\theta_1,\phi_1)$ and $(\theta_2,\phi_2)$ is given by the integral functional  $L[\phi] = R\int^{\theta_{2}}_{\theta_{1}}\sqrt{1+\sin^{2}\theta (\phi')^{2}}d\theta$.



In the problem of finding the shortest path between two points in the plane, we used Cartesian coordinate system and obtained a length integral by taking it over the differential path length along a path
$L[y]=\int^{2}_{1}ds$, where $ds = \sqrt{dx^{2}+dy^{2}}$.

```python
import numpy as np
import matplotlib.pyplot as plt

def sphere_point(r, theta, phi):
    """Convert spherical coordinates (theta=colatitude, phi=azimuth) to Cartesian."""
    return np.array([
        r * np.sin(theta) * np.cos(phi),
        r * np.sin(theta) * np.sin(phi),
        r * np.cos(theta)
    ])

# Parameters
R = 1.0
theta_0 = np.pi / 4      # 45 deg from North pole
phi_0 = np.pi / 4        # 45 deg azimuth
d_theta = 0.18           # Finite differential step in theta
d_phi = 0.22             # Finite differential step in phi

# Base coordinates
P0 = sphere_point(R, theta_0, phi_0)
P_theta = sphere_point(R, theta_0 + d_theta, phi_0)
P_full = sphere_point(R, theta_0 + d_theta, phi_0 + d_phi)

# Tangent vectors at P0 
v_theta = np.array([R * np.cos(theta_0) * np.cos(phi_0),
                    R * np.cos(theta_0) * np.sin(phi_0),
                    -R * np.sin(theta_0)]) * d_theta

v_phi = np.array([-R * np.sin(theta_0) * np.sin(phi_0),
                   R * np.sin(theta_0) * np.cos(phi_0),
                   0.0]) * d_phi

v_ds = v_theta + v_phi
ds_val = np.sqrt((R * d_theta)**2 + (R * np.sin(theta_0) * d_phi)**2)

# Figure setup
fig = plt.figure(figsize=(9, 8))
ax = fig.add_subplot(111, projection='3d')

# 1. Wireframe Sphere
u = np.linspace(0, 2 * np.pi, 36)
v = np.linspace(0, np.pi, 18)
x_sphere = R * np.outer(np.cos(u), np.sin(v))
y_sphere = R * np.outer(np.sin(u), np.sin(v))
z_sphere = R * np.outer(np.ones(np.size(u)), np.cos(v))
ax.plot_wireframe(x_sphere, y_sphere, z_sphere, color='gray', alpha=0.18, linewidth=0.6)

# 2. Base point P0
ax.scatter(*P0, color='crimson', s=45, label=r'Point $P(\theta, \phi)$')

# 3. Tangent vector components (differential triangle)
ax.quiver(*P0, *v_theta, color='teal', linewidth=2.4, arrow_length_ratio=0.15,
          label=r'$R\,d\theta\,\hat{\theta}$')
ax.quiver(*(P0 + v_theta), *v_phi, color='darkorange', linewidth=2.4, arrow_length_ratio=0.15,
          label=r'$R\sin\theta\,d\phi\,\hat{\phi}$')
ax.quiver(*P0, *v_ds, color='crimson', linewidth=3.2, arrow_length_ratio=0.12,
          label=rf'$ds = \sqrt{{R^2 d\theta^2 + R^2 \sin^2\theta d\phi^2}} \approx {ds_val:.3f}$')

# 4. Surface Arc directly connecting P0 to P_full
t_arc = np.linspace(0, 1, 30)
surface_arc = np.array([
    sphere_point(R, theta_0 + t * d_theta, phi_0 + t * d_phi) for t in t_arc
])
ax.plot(surface_arc[:, 0], surface_arc[:, 1], surface_arc[:, 2],
        '--', color='purple', linewidth=2.0, label='Path on Sphere Surface')

# Presentation settings
ax.set_box_aspect([1, 1, 1])
ax.set_xlim([-1.1, 1.1])
ax.set_ylim([-1.1, 1.1])
ax.set_zlim([-1.1, 1.1])
ax.view_init(elev=28, azim=40)
ax.set_title(r'Differential Path Element $ds$ on a 2-Sphere', fontsize=13)
ax.legend(loc='upper left', frameon=True)
plt.show()
```

```python
import numpy as np  # Numerical computing library for trigonometry, arrays, and vector math
import matplotlib.pyplot as plt  # Plotting library for rendering 2D and 3D graphs

def sphere_point(radius, theta, phi):
    """
    Converts spherical coordinates (radius, colatitude theta, azimuth phi)
    to standard 3D Cartesian coordinates [x, y, z].
    
    Arguments:
        radius (float): Distance from the origin to the surface of the sphere.
        theta (float): Colatitude/polar angle in radians, measured from the positive z-axis [0, pi].
        phi (float): Azimuthal angle in radians, measured counterclockwise from the positive x-axis [0, 2*pi).
    
    Returns:
        np.ndarray: 1D NumPy array representing [x, y, z] position vector in Cartesian space.
    """
    x = radius * np.sin(theta) * np.cos(phi)  # Projection onto the x-axis
    y = radius * np.sin(theta) * np.sin(phi)  # Projection onto the y-axis
    z = radius * np.cos(theta)                # Projection onto the z-axis
    return np.array([x, y, z])

# -------------------------------------------------------------------------
# Physical & Differential Parameters
# -------------------------------------------------------------------------
R = 1.0                # Radius of the sphere (arbitrary units)
theta_0 = np.pi / 4    # Base point colatitude: pi/4 rad = 45 degrees from the North Pole (+z)
phi_0 = np.pi / 4      # Base point azimuth: pi/4 rad = 45 degrees from the +x axis in the xy-plane
d_theta = 0.18         # Finite differential step size along the meridian (colatitude direction)
d_phi = 0.22           # Finite differential step size along the parallel (azimuthal direction)

# -------------------------------------------------------------------------
# Reference Coordinate Points on the Spherical Surface
# -------------------------------------------------------------------------
# P0: The base evaluation point r(theta_0, phi_0)
P0 = sphere_point(radius=R, theta=theta_0, phi=phi_0)

# P_theta: The position reached by incrementing only theta: r(theta_0 + d_theta, phi_0)
P_theta = sphere_point(radius=R, theta=theta_0 + d_theta, phi=phi_0)

# P_full: The position reached by incrementing both parameters: r(theta_0 + d_theta, phi_0 + d_phi)
P_full = sphere_point(radius=R, theta=theta_0 + d_theta, phi=phi_0 + d_phi)

# -------------------------------------------------------------------------
# Tangent Vectors and Incremental Displacements
# -------------------------------------------------------------------------
# dr/d_theta: Unscaled partial derivative vector with respect to theta:
# dr_dtheta = [R*cos(theta)*cos(phi), R*cos(theta)*sin(phi), -R*sin(theta)]
# Multiplying by d_theta yields delta_r_theta, the finite displacement vector along the meridian.
delta_r_theta = np.array([
    R * np.cos(theta_0) * np.cos(phi_0),
    R * np.cos(theta_0) * np.sin(phi_0),
    -R * np.sin(theta_0)
]) * d_theta

# dr/d_phi: Unscaled partial derivative vector with respect to phi:
# dr_dphi = [-R*sin(theta)*sin(phi), R*sin(theta)*cos(phi), 0]
# Multiplying by d_phi yields delta_r_phi, the finite displacement vector along the parallel of latitude.
delta_r_phi = np.array([
    -R * np.sin(theta_0) * np.sin(phi_0),
    R * np.sin(theta_0) * np.cos(phi_0),
    0.0
]) * d_phi

# delta_r_ds: The total linear vector displacement (hypotenuse of the differential triangle)
# Formally: dr = (dr/d_theta)*d_theta + (dr/d_phi)*d_phi
delta_r_ds = delta_r_theta + delta_r_phi

# -------------------------------------------------------------------------
# Magnitude Calculations for Path Element ds
# -------------------------------------------------------------------------
# Method 1: Direct Euclidean norm of the composite displacement vector delta_r_ds
ds_from_vector = np.linalg.norm(delta_r_ds)

# Method 2: Analytical metric formula (retained as required):
# ds_val = np.sqrt((R * d_theta)**2 + (R * np.sin(theta_0) * d_phi)**2)

# -------------------------------------------------------------------------
# Figure and 3D Axis Setup
# -------------------------------------------------------------------------
# plt.figure creates the top-level window/canvas.
# 'figsize=(9, 8)' sets the figure dimensions to 9 inches wide by 8 inches tall.
fig = plt.figure(figsize=(9, 8))

# add_subplot(111, projection='3d') creates a single subplot (grid 1x1, position 1)
# and initializes a 3D Cartesian viewing box (Axes3D).
ax = fig.add_subplot(111, projection='3d')

# -------------------------------------------------------------------------
# 1. Wireframe Background Sphere
# -------------------------------------------------------------------------
# np.linspace generates 36 evenly spaced azimuth angles from 0 to 2*pi
u = np.linspace(0.0, 2.0 * np.pi, 36)

# np.linspace generates 18 evenly spaced colatitude angles from 0 to pi
v = np.linspace(0.0, np.pi, 18)

# np.outer calculates the outer product of two vectors to produce coordinate mesh grids:
x_sphere = R * np.outer(np.cos(u), np.sin(v))
y_sphere = R * np.outer(np.sin(u), np.sin(v))
z_sphere = R * np.outer(np.ones(np.size(u)), np.cos(v))

# plot_wireframe draws grid lines across the sphere surface:
# 'color="gray"' sets neutral wire color.
# 'alpha=0.18' makes the wireframe semi-transparent so vectors inside/outside are visible.
# 'linewidth=0.6' sets the thickness of the grid lines.
ax.plot_wireframe(x_sphere, y_sphere, z_sphere, color='gray', alpha=0.18, linewidth=0.6)

# -------------------------------------------------------------------------
# 2. Base Point Evaluation Marker
# -------------------------------------------------------------------------
# scatter plots the reference point P0:
# '*P0' unpacks the 3D coordinates [P0_x, P0_y, P0_z].
# 'color="crimson"' highlights the point.
# 's=45' specifies marker area in points squared.
# 'label' sets the string displayed in the figure legend.
ax.scatter(*P0, color='crimson', s=45, label=r'Base Point $P(\theta_0, \phi_0)$')

# -------------------------------------------------------------------------
# 3. Tangent Differential Triangle Vectors
# -------------------------------------------------------------------------
# Meridional component vector:
# Originates at P0, points in direction of delta_r_theta:
ax.quiver(
    P0[0], P0[1], P0[2],                      # Starting point coordinates (x, y, z)
    delta_r_theta[0], delta_r_theta[1], delta_r_theta[2],  # Vector direction components (u, v, w)
    color='teal',                             # Line color
    linewidth=2.4,                            # Line thickness
    arrow_length_ratio=0.15,                  # Relative size of arrowhead compared to vector length
    label=r'Meridional Displacement $\Delta\mathbf{r}_\theta = R\,d\theta\,\hat{\theta}$'
)

# Azimuthal component vector:
# Originates at the tip of delta_r_theta (P0 + delta_r_theta), displaying vector addition tip-to-tail:
ax.quiver(
    P0[0] + delta_r_theta[0],                 # Shifted start x
    P0[1] + delta_r_theta[1],                 # Shifted start y
    P0[2] + delta_r_theta[2],                 # Shifted start z
    delta_r_phi[0], delta_r_phi[1], delta_r_phi[2],  # Direction components along parallel
    color='darkorange',
    linewidth=2.4,
    arrow_length_ratio=0.15,
    label=r'Azimuthal Displacement $\Delta\mathbf{r}_\phi = R\sin\theta\,d\phi\,\hat{\phi}$'
)

# Total displacement vector (hypotenuse):
# Originates at P0 and points directly to P0 + delta_r_ds:
ax.quiver(
    P0[0], P0[1], P0[2],                      # Starting point coordinates
    delta_r_ds[0], delta_r_ds[1], delta_r_ds[2],  # Combined vector components
    color='crimson',
    linewidth=3.2,
    arrow_length_ratio=0.12,
    label=rf'Hypotenuse $ds = \|\Delta\mathbf{{r}}\| = {ds_from_vector:.3f}$'
)

# -------------------------------------------------------------------------
# 4. True Surface Arc on the Sphere
# -------------------------------------------------------------------------
# Parameter t sweeps linearly from 0 to 1 in 30 increments
t_arc = np.linspace(0.0, 1.0, 30)

# Evaluate the exact spherical positions along the parametric path theta(t), phi(t)
surface_arc = np.array([
    sphere_point(R, theta_0 + t * d_theta, phi_0 + t * d_phi) for t in t_arc
])

# plot renders the curved path lying strictly on the sphere surface:
# '--' specifies a dashed line style to distinguish the surface path from the straight tangent vector ds.
ax.plot(
    surface_arc[:, 0], surface_arc[:, 1], surface_arc[:, 2],
    linestyle='--',
    color='purple',
    linewidth=2.0,
    label='True Path on Curved Sphere Surface'
)

# -------------------------------------------------------------------------
# 5. Scene Formatting, Viewing Angle & Legend
# -------------------------------------------------------------------------
# set_box_aspect([1, 1, 1]) forces equal aspect ratio across x, y, and z axes
# to prevent the sphere from appearing squashed or distorted into an ellipsoid.
ax.set_box_aspect([1, 1, 1])

# set_xlim/ylim/zlim enforce symmetrical bounding box dimensions around the unit sphere
ax.set_xlim([-1.1, 1.1])
ax.set_ylim([-1.1, 1.1])
ax.set_zlim([-1.1, 1.1])

# view_init adjusts the virtual camera perspective:
# 'elev=28' sets the elevation angle 28 degrees above the xy-plane.
# 'azim=40' sets the azimuthal rotation 40 degrees counterclockwise from the x-axis.
ax.view_init(elev=28, azim=40)

# set_title sets the top label of the axes, with 'fontsize=13' specifying text height in points
ax.set_title(r'Differential Path Element $ds$ on a 2-Sphere ($S^2$)', fontsize=13)

# legend generates the explanatory visual box:
# 'loc="upper left"' anchors the legend box in the top-left quadrant of the plot canvas.
# 'frameon=True' renders a background rectangle behind the text for contrast.
ax.legend(loc='upper left', frameon=True)

# plt.show renders the complete OpenGL/interactive canvas to the screen
plt.show()
```
